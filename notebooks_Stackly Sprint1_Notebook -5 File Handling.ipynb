# File Handling

## Reading Text File
Reading a text file means opening an existing file and accessing the information stored inside it. Python provides the open() function to read files. We can use the "r" mode to open a file for reading. For example, if we have a file named sample.txt containing some text, we can use read() to read the complete content of that file.

```python
#Example:

file = open("student.txt","r")
content = file.read()

print(content)
file.close()
```

In this example, open("student.txt", "r") opens the student.txt file in read mode. The statement content = file.read() reads the entire content of the file and stores it in the content variable. The print(content) statement displays the stored text on the screen. Finally, file.close() closes the file after reading. The important point is that read() must have parentheses because they call the method and actually read the file content.

## Writing Text File
Writing a text file means storing information into a file using Python. We use the open() function with "w" mode to write data. If the file already exists, its previous content will be replaced. If the file does not exist, Python creates a new file. For example, we can write a message into a file using the write() method.

```python
#Example:

file = open("student.txt","w")
file.write("How are you")
file.close()
```

## CSV File
A CSV (Comma-Separated Values) file is used to store data in a table format, where each value is separated by a comma. It is commonly used for storing information such as employee details, student records, and sales data. Python provides the built-in csv module to read and write CSV files. For example, we can store student names and marks in a CSV file and read them whenever needed.

```python
#EXample:
# writing a csv

import csv

data = [
    ["Name", "Age", "Marks"],
    ["Naveen", 24, 85],
    ["Rahul", 23, 90]
]

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)
```

```python
# reading CSV file

import csv

with open("students.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

## JSON File
JSON (JavaScript Object Notation) is a format used to store and exchange data in a structured way. It stores information using key-value pairs, similar to a Python dictionary. Python provides the built-in json module to read and write JSON files. For example, we can store a student's name, age, and marks in a JSON file and access that information later.

```python
#Example:

#writing JSON
import json

student = {
    "name": "Naveen",
    "age": 24,
    "marks": 85
}

with open("student.json", "w") as file:
    json.dump(student, file, indent=4)
    
```

```python
# Reading JSON
import json

with open("student.json", "r") as file:
    data = json.load(file)

print(data)
print(data["name"])
```

## Pickle Files
Pickle is a Python module used to store Python objects in a file and retrieve them later. It can save different Python data types, such as lists, dictionaries, and objects, without converting them manually into text. For example, we can store a Python dictionary in a pickle file and load it again whenever required. Pickle files are mainly used when we want to save Python objects in their original structure.

```python
#Writing into Pickle file

import pickle

student = {
    "name": "Naveen",
    "age": 24,
    "marks": 85
}

with open("student.pkl", "wb") as file:
    pickle.dump(student, file)
```

```python
# Reading Pickle file

import pickle

with open("student.pkl", "rb") as file:
    data = pickle.load(file)

print(data)
```

## Context Manager
A Context Manager is used to manage resources automatically, such as opening and closing files. The with statement makes file handling easier because we do not need to close the file manually. Once the code inside the with block is completed, Python automatically closes the file, even if an error occurs. For example, we can use with open() to read or write a file safely.

```python
#Example:
with open("student.txt", "r") as file:
    content = file.read()
    print(content)
```

In this example, with opens the file and allows us to read its content. After the code inside the block finishes, Python automatically closes the file. This makes the code simpler and helps prevent files from remaining open accidentally.

## pathlib
pathlib is a Python module used to work with file and directory paths. It provides an easy and readable way to create, access, and manage paths without manually joining folder names and file names. For example, we can use Path() to represent a file path and check whether the file exists.

```python
#Example:
from pathlib import Path

file_path = Path("student.txt")

if file_path.exists():
    print("File exists")
else:
    print("File does not exist")
```

In this example, Path() creates a path object for sample.txt. The exists() method checks whether the file or directory is available at that location. If it exists, the program prints "File exists"; otherwise, it prints "File does not exist".

## os
The os module is used to interact with the operating system through Python. It helps us perform operations such as checking the current working directory, listing files, creating folders, and accessing environment information. For example, we can use os.getcwd() to find the current working directory of our Python program.

```python
#Example:
import os

current_directory = os.getcwd()

print("Current Directory:", current_directory)
```

In this example, os.getcwd() returns the location of the directory where the Python program is currently running. The returned path is stored in current_directory and displayed using print().

## Shutil
shutil is a Python module used to perform high-level file and directory operations. It helps us copy, move, and delete files or folders easily. For example, we can use shutil.copy() to create a copy of an existing file at another location.

```python
#Example:
import shutil

shutil.copy("student.txt", "student_copy.txt")

print("File copied successfully")
```

In this example, shutil.copy() copies the content of sample.txt into a new file named sample_copy.txt. The original file remains unchanged, and a separate copy is created.

## File validation
File Validation means checking whether a file meets the required conditions before using it. We can validate whether a file exists, whether it is actually a file, and whether it has the correct extension. For example, before reading a CSV file, we can check whether the file exists and whether its extension is .csv. This helps prevent errors while working with files.

```python
#Example:
from pathlib import Path

file_path = Path("students.csv")

if file_path.exists() and file_path.is_file() and file_path.suffix == ".csv":
    print("Valid CSV file")
else:
    print("Invalid or missing file")
```

In this example, exists() checks whether the path is available, is_file() checks whether it is a file, and suffix checks its extension. If all three conditions are true, the program displays "Valid CSV file".

## Directory operations
Directory Operations are used to create, view, and manage folders using Python. A directory is another name for a folder. Python provides modules such as pathlib and os to perform directory operations. For example, we can create a new folder using mkdir() and list the files inside a folder using iterdir().

```python
#Example:
from pathlib import Path

folder = Path("MyFolder")

# Create a directory
folder.mkdir(exist_ok=True)

# Display items inside the directory
for item in folder.iterdir():
    print(item)

print("Directory created or already exists")
```

In this example, Path("MyFolder") represents a folder named MyFolder. The mkdir() method creates the folder, and exist_ok=True prevents an error if the folder already exists. The iterdir() method displays the items available inside that folder.
