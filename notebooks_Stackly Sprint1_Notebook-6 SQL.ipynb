# SQL

SQL (Structured Query Language) is a language used to communicate with databases and we use it to create tables, insert data, retrieve data, update records, and delete records.

## Database

A Database is an organized collection of data that is stored and managed so that we can easily access, update, and use the information. For example, a college can have a database that stores student details, department details, marks, and attendance.

```python
#Example :

%%sql

create table Students (
    student_id INT,
    name varchar,
    marks INT);
```

Here, we are creating a table called Students to store student information. The table is part of a database and contains columns for student ID, name, and marks.

## tables
A Table is used to store data in rows and columns inside a database. Each row represents one record, and each column represents a particular type of information.

```python
%%sql

CREATE TABLE Students (
    student_id INTEGER,
    name TEXT,
    age INTEGER,
    marks INTEGER
);
```

Here, Students is the table. student_id, name, age, and marks are columns. Each student stored in the table will have one row.

## primary key
A Primary Key is a column that uniquely identifies each record in a table. A primary key cannot contain duplicate or NULL values. For example, every student can have a unique student ID.

```python
%%sql

CREATE TABLE Students_1 (
    student_id INTEGER PRIMARY KEY,
    name TEXT,
    marks INTEGER
);
```


Here, student_id is the primary key. Every student must have a different student_id.

## Foreign Key

A Foreign Key is a column that connects one table to another table. It usually refers to the primary key of another table.


```python
%%sql

CREATE TABLE Departments (
    department_id INTEGER PRIMARY KEY,
    department_name TEXT
);

CREATE TABLE Students (
    student_id INTEGER PRIMARY KEY,
    name TEXT,
    department_id INTEGER,
    FOREIGN KEY (department_id)
        REFERENCES Departments(department_id)
);
```

department_id in the Students table is a foreign key. It connects the student with a department from the Departments table.

## Relationships
A Relationship describes how data in two or more tables is connected. Common relationships are one-to-one, one-to-many, and many-to-many. For example, one department can have many students, which is a one-to-many relationship.


```python
%%sql

CREATE TABLE Courses (
    course_id INTEGER PRIMARY KEY,
    course_name TEXT,
    department_id INTEGER,
    FOREIGN KEY (department_id)
        REFERENCES Departments(department_id)
);
```

## Constraints
Constraints are rules applied to table columns to control what type of data can be stored. Common constraints are PRIMARY KEY, FOREIGN KEY, NOT NULL, UNIQUE, CHECK, and DEFAULT.

```python
%%sql

CREATE TABLE Students_2 (
    student_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    email TEXT UNIQUE,
    age INTEGER CHECK(age >= 18)
);
```

## select
SELECT is used to retrieve data from a database table. We can use it to display specific columns or all columns from a table. It helps us choose exactly what information we want to see in the query result.

```python
%%sql 
select employee_name,employee_id from Employees;
```

## where
WHERE is used to filter records from a table based on one or more specified conditions. It returns only the rows that satisfy the given condition and ignores the remaining rows.

```python
%%sql

SELECT *
FROM Employees
WHERE salary > 45000;
```

## Order by

ORDER BY is used to arrange the records returned by a query in a specific order. We can sort the data in ascending order using ASC or descending order using DESC. By default, it sorts the data in ascending order.

```python
%%sql
SELECT *
FROM Employees
ORDER BY salary DESC;
```

## Group by
GROUP BY is used to combine rows that have the same values in one or more columns into groups. It is commonly used with aggregate functions such as COUNT(), SUM(), AVG(), MIN(), and MAX() to perform calculations on each group.

```python
%%sql

SELECT department_id, COUNT(*) AS total_employees
FROM Employees
GROUP BY department_id;
```

## Having
HAVING is used to filter the results created by the GROUP BY clause. It applies conditions to groups rather than individual rows and is commonly used with aggregate functions such as COUNT(), SUM(), and AVG().

```python
%%sql

SELECT department_id, COUNT(*) AS total_employees
FROM Employees
GROUP BY department_id
HAVING COUNT(*) > 1;
```

## Distinct
DISTINCT is used to remove duplicate values from the query result and return only unique records. It can be applied to one or more columns, helping us identify different values available in a table.

```python
%%sql

SELECT DISTINCT department_id
FROM Employees;
```

## limt

LIMIT is used to restrict the number of records returned by a SELECT query. It is useful when we want to display only a specific number of rows instead of retrieving the entire table.

```python
%%sql

SELECT *
FROM Employees
LIMIT 3;
```

# Joins

## Inner Join
INNER JOIN is used to combine records from two or more tables based on a matching condition between their related columns. It returns only the rows where matching values are available in both tables. Records without a match are excluded from the result.

```python
%%sql

SELECT Employees.employee_name, Departments.department_name
FROM Employees
INNER JOIN Departments
ON Employees.department_id = Departments.department_id;
```

## Left Join
LEFT JOIN is used to return all records from the left table and only the matching records from the right table based on the specified condition. If a record from the left table does not have a match in the right table, the right table columns will contain NULL values.

```python
%%sql
SELECT Employees.employee_name, Departments.department_name
FROM Employees
LEFT JOIN Departments
ON Employees.department_id = Departments.department_id;
```

## Right Join
RIGHT JOIN is used to return all records from the right table and only the matching records from the left table. If a record from the right table does not have a matching record in the left table, the left table columns will contain NULL values.

```python
%%sql
SELECT Employees.employee_name, Departments.department_name
FROM Employees
RIGHT JOIN Departments
ON Employees.department_id = Departments.department_id;
```

## Full Outer Join
FULL OUTER JOIN is used to combine records from two tables and return all rows from both tables. It includes matching records as well as non-matching records from either table. When a record has no match, the columns from the other table will contain NULL values.

```python
%%sql
SELECT Employees.employee_name, Departments.department_name
FROM Employees
FULL OUTER JOIN Departments
ON Employees.department_id = Departments.department_id;
```

## Cross Join

CROSS JOIN is used to combine every row from the first table with every row from the second table. It produces all possible combinations of records between the two tables, without requiring a matching condition.


```python
%%sql
SELECT Employees.employee_name, Departments.department_name
FROM Employees
CROSS JOIN Departments;
```

# Advanced Sql

## union

UNION is used to combine the results of two or more SELECT queries into a single result set. It removes duplicate rows from the final result. All SELECT statements must return the same number of columns, and the corresponding columns must have compatible data types.

```python
%%sql

SELECT employee_name AS name
FROM Employees

UNION

SELECT department_name AS name
FROM Departments;
```

## Case

CASE is used to apply conditional logic inside a SQL query. It checks conditions one by one and returns a specified value when a condition is satisfied. It works like an if-elif-else statement in programming and is commonly used to create categories or labels.

```python
%%sql

SELECT
    employee_name,
    salary,
    CASE
        WHEN salary >= 50000 THEN 'High Salary'
        WHEN salary >= 40000 THEN 'Medium Salary'
        ELSE 'Low Salary'
    END AS salary_category
FROM Employees;
```

## Subqueries
A subquery is a SQL query written inside another SQL query. It is used to retrieve data that can be passed to the main query as a condition or result. Subqueries are commonly used with SELECT, WHERE, and FROM clauses to solve problems that require multiple steps.

```python
%%sql
SELECT employee_name, salary
FROM Employees
WHERE salary > (
    SELECT AVG(salary)
    FROM Employees
);
    
```

## CTE (Common Table Expression)
A CTE is a temporary named result set created using the WITH keyword. It allows us to write a query in smaller, organized sections and reuse the result within the main query. CTEs make complex SQL queries easier to read, understand, and maintain.

```python
%%sql

WITH AverageSalary AS (
    SELECT AVG(salary) AS avg_salary
    FROM Employees
)
SELECT employee_name, salary
FROM Employees
WHERE salary > (
    SELECT avg_salary
    FROM AverageSalary
);
```

## views
A view is a virtual table created from a SQL query. It does not usually store the actual data separately; instead, it displays data retrieved from one or more underlying tables whenever the view is queried. Views help simplify complex queries, organize data access, and show only selected columns or records.

```python
%%sql
CREATE VIEW HighSalaryEmployees AS
SELECT employee_id, employee_name, salary
FROM Employees
WHERE salary >= 50000;
```

```python
%%sql
SELECT *
FROM HighSalaryEmployees;
```

## indexes
An index is a database object created on one or more columns to help the database find records more quickly. It works similarly to an index in a book, allowing the database to locate required data without checking every row. However, indexes require additional storage and may slow down INSERT, UPDATE, and DELETE operations because the index also needs to be maintained.

```python
%%sql
CREATE INDEX idx_employee_name
ON Employees(employee_name);
```

# Window Functions

## Row Number()
ROW_NUMBER() is a window function that assigns a unique sequential number to each row within a result set. The numbering is determined by the ORDER BY clause inside the OVER() function. Even when two rows have the same value, they receive different row numbers.

```python
%%sql
SELECT
    employee_name,
    salary,
    ROW_NUMBER() OVER (ORDER BY salary DESC) AS row_num
FROM Employees;
```

## Rank()
RANK() is a window function used to assign rankings to rows based on the specified column values. Rows with equal values receive the same rank, but the next rank is skipped after a tie. It is useful when ranking employees based on salary or performance.

```python
%%sql
SELECT
    employee_name,
    salary,
    RANK() OVER (ORDER BY salary DESC) AS employee_rank
FROM Employees;
```

## Dense_Rank()
DENSE_RANK() is a window function used to assign rankings to rows based on the specified column values. Similar to RANK(), it gives the same rank to rows with equal values, but it does not skip the next rank after a tie. This means the ranking continues without gaps.

```python
%%sql
SELECT
    employee_name,
    salary,
    DENSE_RANK() OVER (ORDER BY salary DESC) AS employee_rank
FROM Employees;
```

## Lag()
LAG() is a window function used to access a value from a previous row in the result set without using a self-join. It retrieves the value based on the ordering specified inside the OVER() clause. It is useful for comparing the current row with a previous row, such as comparing employee salaries.

```python
%%sql
SELECT
    employee_name,
    salary,
    LAG(salary) OVER (ORDER BY employee_id) AS previous_salary
FROM Employees;
```

## Lead()
LEAD() is a window function used to access a value from a following row in the result set without using a self-join. It retrieves the value based on the ordering specified inside the OVER() clause. It is useful for comparing the current row with the next row or analyzing changes between consecutive records.

```python
%%sql
SELECT
    employee_name,
    salary,
    LEAD(salary) OVER (ORDER BY employee_id) AS next_salary
FROM Employees;
```
