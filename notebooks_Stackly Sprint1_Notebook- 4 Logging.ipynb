# Logging

Logging is used to record what is happening inside a Python program.Instead of only displaying messages on the screen, logging allows us to record messages with information such as time, severity, and where the message came from.

```python
#Example:

import logging

logging.basicConfig(level = logging.info)

logging.info("Pragramm Started")
logging.warning("No storage Space")
```

In this example, we import the logging module and set the minimum logging level to INFO. The program then records three messages: one showing that the program started, another warning about low storage space. These messages are displayed in the console by default.

## Why use logging

Logging helps us understand what happened in a program, especially when something goes wrong. It is useful for debugging, tracking program activities, identifying errors, and keeping records for future reference.

For example, if a program stops working, we can check its logs to find out what happened before the error occurred.

Logging can also save messages to a file, allowing us to review them later.

# Logging Levels

## DEBUG

DEBUG is the lowest standard logging level. It is used for detailed technical information that helps developers understand exactly what the program is doing internally. Debug messages are mainly useful during development, testing, and troubleshooting. They can contain information about variable values, function execution, program flow, database queries, or other internal operations. These messages are usually not necessary for normal users.

In simple words:

DEBUG = Detailed information used to find and fix problems.

## Info

INFO is used to record normal and expected events in a program. It tells us that an important operation has happened successfully or that the program is working normally. Examples include starting an application, a user logging in, completing a transaction, or successfully connecting to a service. INFO messages are generally useful for monitoring the normal operation of an application.

In simple words:

INFO = Information about normal program activi

## warning

WARNING indicates that something unusual or potentially problematic has happened, but the program can still continue working. It is useful when the current situation is not necessarily an error but could cause a problem in the future. Examples include low storage space, an approaching limit, an expired configuration, or using a default setting. A warning tells developers or administrators that they may need to pay attention.

In simple words:

WARNING = Something needs attention, but the program can continue.

## Error
ERROR indicates that a particular operation has failed or something went wrong while the program was running. The application may still be able to continue, but the specific operation could not be completed successfully. Errors can occur because of invalid data, failed database operations, network problems, missing files, or other unexpected situations. Error logs are especially useful for troubleshooting application problems.

In simple words:

ERROR = Something went wrong and an operation failed.

## Critical

CRITICAL is the highest standard logging level. It represents a very serious problem that may prevent the application or an important part of the system from continuing to work correctly. Critical messages are generally reserved for major failures that require immediate attention. Examples can include a completely unavailable database, failure of a core service, or a serious system-level problem.

In simple words:

CRITICAL = A very serious problem that may affect the entire application.

```python
#Example:

import logging

logging.basicConfig(level=logging.DEBUG)

username = "Naveen"
balance = 5000
withdraw = 3000

logging.debug("Checking account details")

logging.info("User logged in successfully")

if balance < 1000:
    logging.warning("Account balance is low")

if withdraw > balance:
    logging.error("Withdrawal failed: insufficient balance")
else:
    balance = balance - withdraw
    logging.info("Withdrawal successful")

if balance < 100:
    logging.critical("Account balance is critically low")

logging.debug(f"Remaining balance: {balance}")
```

## Log Formatters
A Log Formatter decides what information is shown in each log message and in what order. For example, we can include the time, log level, and actual message.

```python
#Example:
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s", force = True)

logging.info("User logged in")
```

Here, asctime shows the time, levelname shows the level such as INFO, and message shows the actual log message.

## File Handlers
A File Handler is used to save log messages into a file instead of showing them only on the screen.It is useful when we want to store logs and check them later.


```python
#Example:
import logging

logging.basicConfig(
    filename="app.log",
    level=logging.INFO)

logging.info("Application started")
logging.warning("Low memory")
logging.error("Something went wrong")
```

Here in the example, the programm created a file and stores the data in it, later we can access the data from it



## Console Handlers
A Console Handler is used to display log messages directly in the console or terminal while the program is running. It helps us monitor what is happening in the application in real time. For example, if we want to know whether the application has started successfully or if an error has occurred, we can use a Console Handler to display those messages on the screen. It is mainly useful for debugging and monitoring an application.

```python
#Example:
import logging

logger = logging.getLogger("MyApp")

console_handler = logging.StreamHandler()
logger.addHandler(console_handler)

logger.setLevel(logging.INFO)

logger.info("Application started")
logger.warning("Low memory")
logger.error("Something went wrong")
```

In this example, StreamHandler() creates a Console Handler and addHandler() connects it to the logger. When the info(), warning(), and error() methods are used, their messages are displayed directly in the console. This allows us to see the application's activities while it is running without saving the messages to a file.

## Rotating Logs - Bonus
A log file can become very large if an application runs for a long time. Rotating Logs solve this problem by automatically creating a new log file when the current file reaches a specified size.

```python
#Example:
import logging
from logging.handlers import RotatingFileHandler

handler = RotatingFileHandler(
    "app.log",
    maxBytes=1000000,
    backupCount=3)

logging.basicConfig(
    level=logging.INFO,
    handlers=[handler])

logging.info("Application started")
```

Here, maxBytes=1000000 means the log file can reach approximately 1 MB before rotation occurs. backupCount=3 means Python keeps up to three older log files. This prevents one log file from growing indefinitely.

## Logging Best Practices
Logging Best Practices are simple rules we should follow when using logging in a Python application. Good logging helps us understand what is happening in the application, find errors easily, and troubleshoot problems without changing the code. We should use the correct log level for each message, such as INFO for normal activities, WARNING for possible problems, ERROR for errors, and CRITICAL for serious problems. We should write clear and meaningful log messages instead of unnecessary messages. For larger applications, it is useful to save logs in a file so they can be checked later. We should also avoid storing sensitive information such as passwords, bank details, or personal data in logs.

```python
#Example:
import logging

logging.basicConfig(level=logging.INFO)

logging.info("User logged in")
logging.warning("Low account balance")
logging.error("Payment failed")
```

In this example, INFO is used for a normal activity because the user successfully logged in. WARNING is used because the account balance is low and may need attention. ERROR is used because the payment failed and something went wrong. Using the correct log level makes the logs easier to understand and helps developers quickly identify problems.
