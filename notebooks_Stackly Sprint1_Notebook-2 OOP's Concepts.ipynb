# Object-Oriented Programming

OOP's is an programming approach where the programs are organized using classes and objects.

## Class

A class is like a blueprint or template used to create objects. It defines the attributes (data) and methods (behavior) that objects created from it will have. Instead of writing the same code repeatedly for similar entities, a class allows us to define the structure once and create multiple objects whenever needed.


```python
#Example:

class Student:
    name = "Naveen"
    age = 24

```

Here Student is the class, name and age are the information inside it

## Object
An object is an actual instance created from a class. If a class is a blueprint, an object is the actual thing created using that blueprint.

```python
#Example:

class Student:
    name = "Naveen"
    age = 24


student1 = Student()

print(student1.name)
print(student1.age)
```

Here student1 is the object, and it can access the information inside the Student class.
Student → class
student1 → object
Student() → creates the object

## Constructor  __init__()

A Constructor is a special method in Python that is automatically executed when an object is created. It is defined using the __init__ method and is mainly used to initialize the attributes of an object with their initial values.

```python
#Example:

class Student:

    def __init__(self, name):
        self.name = name
student1 = Student("Naveen")

print(student1.name)
```

Here in the code we created a Student class and uses it to create a student object. The __init__() method runs automatically when we create the object. Here, name receives "Naveen", and self.name = name stores "Naveen" inside the student1 object. So, student1 now has a name called "Naveen". Finally, print(student1.name) gets the name stored in student1 and prints Naveen.

## Instance Variables
An instance variable is a variable that belongs to a particular object and each object can have its own separate values.

```python
#Example:

class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Naveen", 24)
student2 = Student("Rahul", 22)

print(student1.age)
print(student2.name)
```

Here in the example, name and age are instance variables because they store separate values for each student object. student1 stores "Naveen" and 24, while student2 stores "Rahul" and 22. The self refers to the current object, so self.name stores the name inside that particular student object.

## Class Variables
A class variable  are variables that belong to the class itself rather than to individual objects. They are declared directly inside the class but outside any methods. Since they are associated with the class, all objects created from that class share the same variable.

```python
#Example:

class Student:

    school = "ABC School"

    def __init__(self, name,age):
        self.name = name
        self.age = age

student1 = Student("Naveen",24)
student2 = Student("Rahul",27)

print(student1.school)
print(student2.school)
```

Here in the example, school is a class variable because it is defined directly inside the Student class and is shared by all student objects. Both student1 and student2 can access the same "ABC School" value, while name is an instance variable because each student has a different name.

## Instance Methods
An instance method is a function inside a class that works with the data of a particular object. It uses self to access that object's instance variables.

```python
#Example:

class Student:

    def __init__(self, name):
        self.name = name
    def introduce(self):
        print("My name is", self.name)
student1 = Student("Naveen")

student1.introduce()
```

Here in the example, introduce() is an instance method because it uses self to access the name of the particular student object. When we call student1.introduce(), self refers to student1, so self.name gives "Naveen".

## Static Methods
A static method is a method inside a class that does not need object data and it does not use self.

```python
#Example:

class Student:
    @staticmethod
    def college_info():
        print("College")

Student.college_info()
```

Here in the example, college_info() is a static method because it does not need any student object or object data. It simply performs an independent task, so we can call it directly using Student.college_info() without creating an object.

## Class Methods
A class method is a method that works with the class itself, rather than with a particular object and We use @classmethod and cls inside it.

```python
#Example:

class Student:

    school = "Vidhyapeet"
    @classmethod
    def show_school(cls):
        print(cls.school)

Student.show_school()
```

Here in the example, show_school() is a class method because it works with the class variable school. The cls refers to the Student class, so cls.school gets the value Vidyapeet. We can call the method using Student.show_school() without creating a student object.

## Properties
A property allows us to access a method like a normal variable and We use @property when we want to control how a value is accessed.

```python
#Example:

class Student:

    def __init__(self, name):
        self.name = name

    @property
    def student_name(self):
        return self.name
student1 = Student("Naveen")

print(student1.student_name)
```

Here in the example, student_name() is converted into a property using @property, so we can access it like a normal variable using student1.student_name instead of calling it with parentheses. The property gets the student's name from self.name and returns it.

## Encapsulation
Encapsulation means keeping an object's data and the methods that work with that data together inside a class and controlling how the data is accessed.

```python
#example:

class Employee:

    def __init__(self, salary):
        self.__salary = salary

    def show_salary(self):
        print(self.__salary)
employee1 = Employee(30000)

employee1.show_salary()
```

Here in the example, __salary is kept inside the Employee class, so the salary is protected from direct access outside the class. The show_salary() method is used to access the salary in a controlled way. This is called encapsulation because we keep the data and the method that works with it together inside the class.

## Abstraction
Abstraction is one of the core principles of Object-Oriented Programming (OOP). It is the process of hiding the internal implementation details of a class while exposing only the essential features required by the user. This allows users to interact with an object without needing to understand how the underlying functionality is implemented.

```python
#Example:

class Mobile:

    def make_call(self, number):
        print("Calling", number)
mobile = Mobile()

mobile.make_call("1234567890")
```

Here in the example, we simply use make_call() to make a call. We don't need to know how the mobile connects to the network, finds the tower, or sends the signal. All those internal details are hidden, and we only use the simple make_call() method.

## Inheritance
Inheritance allows one class to use properties and methods from another class.The existing class is called the parent/base class and the new class is called the child/derived class.

```python
#Example:

class Animal:
    def eat(self):
        print("Animal is eating")

class Dog(Animal):
    def bark(self):
        print("Dog is barking")

dog = Dog()

dog.eat()
dog.bark()
```

Here in the example, Animal is the parent class and Dog is the child class. The Dog class inherits the eat() method from Animal, so we can use dog.eat() even though eat() is not written inside the Dog class. The Dog class also has its own bark() method. This allows us to reuse code from the parent class instead of writing it again.

## Multiple Inheritance
Multiple inheritance is a type of inheritance in which a single child class inherits attributes and methods from two or more parent classes. This allows the child class to combine the functionalities of multiple classes into one, enabling greater code reuse and flexibility.

```python
#Example:

class Printer:

    def print_document(self):
        print("Printing document")

class Scanner:

    def scan_document(self):
        print("Scanning document")

class AllInOne(Printer, Scanner):

    def copy_document(self):
        print("Copying document")

device = AllInOne()

device.print_document()
device.scan_document()
device.copy_document()
```

Here in the example, Printer and Scanner are two parent classes, and AllInOne is the child class. AllInOne inherits features from both classes, so it can print, scan, and also perform its own copy operation. This is called multiple inheritance because one class gets features from more than one parent class.

## Multilevel inheritance
Multilevel Inheritance is a type of inheritance in which a class inherits from another derived class, forming a chain of inheritance. In this hierarchy, one class acts as both a child of its parent class and a parent to another child class.

```python
#Example:

class Animal:
    def eat(self):
        print("Animal can eat")
class Dog(Animal):

    def bark(self):
        print("Dog can bark")
class Puppy(Dog):

    def play(self):
        print("Puppy can play")
puppy = Puppy()

puppy.eat()
puppy.bark()
puppy.play()
```

Here in the example, Animal is the parent of Dog, and Dog is the parent of Puppy. So Puppy gets features from both levels: it can use bark() from Dog and eat() from Animal. It also has its own play() method. This is called multilevel inheritance because inheritance happens through multiple levels.

## Hierarchical Inheritance
Hierarchical inheritance is a type of inheritance where multiple child classes inherit properties and methods from the same parent class. The parent class contains common features that can be shared by all the child classes, while each child class can also have its own additional features.

```python
#Example:

class Vehicle:

    def start(self):
        print("Vehicle is starting")
class Car(Vehicle):

    def drive(self):
        print("Car is driving")
class Bike(Vehicle):

    def ride(self):
        print("Bike is riding")

car = Car()
bike = Bike()

car.start()
car.drive()

bike.start()
bike.ride()
```

Here in the example, Vehicle is the parent class, and both Car and Bike are child classes. Both child classes inherit the start() method from Vehicle, while each child also has its own method. This is called hierarchical inheritance because multiple child classes inherit from the same parent class.

## Polymorphism
Polymorphism means one method or function can behave differently depending on the object or situation.

```python
#Example:

class Dog:

    def sound(self):
        print("Dog says Woof")
class Cat:
    def sound(self):
        print("Cat says Meow")

dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

Here in the example, both Dog and Cat classes have the same method name, sound(), but the method behaves differently for each object. When we call dog.sound(), it prints "Dog says Woof", while cat.sound() prints "Cat says Meow".

## Method Overriding
Method Overriding is an Object-Oriented Programming concept in which a child class provides its own implementation of a method that is already defined in its parent class. The child class uses the same method name, the same parameters, and the same return type (where applicable), but changes the behavior according to its own requirements.

```python
#Example:

class Animal:
    def sound(self):
        print("Animal makes a sound")
class Dog(Animal):
    def sound(self):
        print("Dog says Woof")

dog = Dog()

dog.sound()
```

Here in the example, the Animal class has a sound() method, and the Dog class inherits from Animal but creates its own sound() method with the same name. When we call dog.sound(), Python uses the Dog version instead of the Animal version. This is called method overriding because the child class provides its own version of the parent's method.

## Method Overloading — Python Approach
Method overloading means having the same method name with different numbers or types of arguments.In languages like Java, we can directly create multiple methods with the same name. Python does not support traditional method overloading.

```python
#Example:
class Calculator:

    def add(self, a, b=0, c=0):
        return a + b + c
calculator = Calculator()

print(calculator.add(10))
print(calculator.add(10, 5))
print(calculator.add(10, 5, 15))
```

Here in the example, we have only one add() method, but it can work with one, two, or three values because b and c have default values of 0. When we provide different numbers of arguments, the same method performs the addition accordingly. This is how Python achieves behavior similar to method overloading.

## Composition
Composition is an OOP concept where one class contains an object of another class and uses it to perform some work.

```python
#Example:
class Engine:
    def start(self):
        print("Engine started")

class Car:
    def __init__(self):
        self.engine = Engine()
    def drive(self):
        self.engine.start()
        print("Car is driving")

car = Car()
car.drive()
```

Here in the example, Engine and Car are two separate classes. Inside the Car class, we create an Engine object using self.engine = Engine(). This means the Car has an Engine and uses it when the drive() method is called. This relationship between objects is called composition.

## Aggregration
Aggregation is a relationship between two classes where one class has or uses an object of another class, but the objects can still exist independently. The contained object is created separately and then passed to the other class.

```python
#example:

class Employee:
    def work(self):
        print("Comapny Employee")
class Company:
    def __init__(self, employee):
        self.employee = employee

    def show_employee(self):
        self.employee.work()
        print("Work every day")
employee1 = Employee()
company1 = Company(employee1)

company1.show_employee()
```

Here in the example, Employee and Company are two separate classes. The employee1 object is created independently and then passed to the Company class. The company uses the employee object through self.employee to call the work() method. Even if the Company object is removed, the Employee object can still exist independently. This relationship is called aggregation.

## Association
Association is a relationship between two separate classes where their objects interact with or use each other, but neither object owns the other. Both objects can exist independently.

```python
#Example:
class Teacher:
    def teach(self):
        print("Teacher is teaching")

class Student:
    def learn(self):
        print("Student is learning")

teacher1 = Teacher()
student1 = Student()

teacher1.teach()
student1.learn()
```

Here in the example, Teacher and Student are two separate classes. Their objects are created independently, and they can exist without depending on each other. They are related because a teacher teaches and a student learns, but neither class owns the other. This relationship between the two classes is called association.

## Magic (Dunder) Methods
Magic methods, also called dunder methods, are special methods in Python that have double underscores (__) before and after their names. They allow Python objects to behave in a special way when we perform operations like creating an object, adding objects, comparing objects, printing objects, etc.

```python
#Example:

class Student:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name
student1 = Student("Naveen")

print(student1)
```

Here in the example, __init__() is used to set the student's name when the object is created. The __str__() method controls what happens when we use print(student1). Instead of displaying the default object information, Python calls __str__() and displays the student's name. These special methods with double underscores are called magic or dunder methods.

## Super()
super() is used in inheritance to access methods or variables from the parent class inside the child class. It is commonly used when the child class wants to reuse the parent's constructor or method instead of writing the same code again.

```python
#Example:

class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        super().sound()
        print("Dog says Woof")

dog = Dog()
dog.sound()
```

Here in the example, Dog is the child class and Animal is the parent class. The Dog class has its own sound() method, but super().sound() is used to call the sound() method from the parent class first. After that, the child class prints its own message.

## __str__()
__str__() is a magic (dunder) method in Python that controls what should be displayed when we print an object.Without __str__(), printing an object usually gives a default output containing the class name and memory-related information. By defining __str__(), we can make the output simple and meaningful.

```python
#Example:
class Student:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return self.name

student1 = Student("Naveen")

print(student1)
```

Here in the example, Student is a class with a name variable. The __str__() method returns the student's name whenever the object is printed using print(student1). Instead of showing Python's default object information, it displays Naveen, making the output easier to understand.

## __repr__()
__repr__() is a magic (dunder) method used to define the official or developer-friendly representation of an object.It is mainly useful when we want to see detailed information about an object while debugging or developing a program.

```python
#Example:
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"Student(name='{self.name}', age={self.age})"

student1 = Student("Naveen", 24)

print(repr(student1))
```

Here in the example, __repr__() defines how the Student object should be represented when we use repr(student1). Instead of showing Python's default object information, it displays useful details such as the student's name and age. This makes the object easier to understand when debugging or checking program output.

## __len__()
__len__() is a magic (dunder) method used to define what should happen when we use the built-in len() function on an object.It tells Python how many items or elements an object contains.

```python
#Example:
class Team:
    def __init__(self, players):
        self.players = players

    def __len__(self):
        return len(self.players)

team1 = Team(["Naveen", "Rahul", "Arun"])

print(len(team1))
```

Here in the example, Team contains a list of three players. The __len__() method tells Python to count the items inside the players list. When we use len(team1), Python automatically calls __len__() and returns 3.

## Dataclasses — Bonus
Dataclasses are a simple way to create classes that are mainly used for storing data.Normally, when creating a class to store data, we have to write methods like __init__() ourselves. With Python's @dataclass, Python automatically creates common methods for us, which makes the code shorter and easier to read.

```python
#Example:
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    age: int
    course: str
student1 = Student("Naveen", 24, "Python")

print(student1)
```

Here in the example, @dataclass is used to create a class for storing student information. We don't need to manually write the __init__() method because the dataclass creates it automatically. We can directly create student1 by providing the name, age, and course, and Python also provides a useful string representation when we print the object.

## Singleton Pattern — Bonus
The Singleton Pattern is a design pattern that makes sure a class has only one object (instance) throughout the program.If we try to create the object multiple times, Python will return the same object instead of creating a new one.

A common use case is a database connection, configuration manager, or logging system, where we may want only one shared instance.

```python
#Example:
class Database:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance
db1 = Database()
db2 = Database()

print(db1 is db2)
```

Here in the example, Database is designed to create only one object. When db1 is created, Python creates the first object and stores it in _instance. When db2 is created, Python checks _instance and returns the existing object instead of creating a new one. Therefore, db1 and db2 refer to the same object.
