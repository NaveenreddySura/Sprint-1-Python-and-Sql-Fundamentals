Control Flow

If/Else/Elif

If/Elif/Else are used in program to take decisions

if - checks the condition Elif - checks another condition Else - if both conditions fails Else will takes place

Example:

marks = int(input("Enter the marks")) 
if marks>80:
    print("A grade")
elif marks>60:
    print("B grade")
else: print("Fail")

In this programm first it checked the "if" condition if it is true it will print the IF statement if it is false the it will go to "Elif" condition if Elif is false then it will go to Else condition at last

For Loop

For loop is used to repeat the block of code for each values in sequence

Example:

names = ["naveen", "Praveen", "Kumar"] 
for name in names:
    print("Hello",name)

Here output will be Hello Naveen, Hello Praveen, Hello Kumar loop will repeat the same "Hello" code to every value in sequence

While loop is uded to ececuted block of code repeatedly until the condition is true

Example:

number = 1
while number<=5:
    print("Number is",number)
    number = number+1

In the programm it repeated the code again and again until it condition is true, it returned until 1,2,3,4,5 then it stoped becuase we give the condtion print until 5 number

Break

break is used to immediately Stop the loop evn it is not finished

example:

marks = [60,70,80,90,100]

for mark in marks:
    if mark == 90:
        break
    print(mark)

In the program we given values to the variable 60,70,80,90,100 it is a loop so it will print the all number but also we have the break condtion so if it matches the breaks condtion that break condition will stop the loop so we got output only 60,70

Continue

continue is used to skip the current iteration of a loop and move to the next iteration and it doesnt stop the loop

Example:

numbers = [50,60,70,80,90]

for number in numbers:
    if number == 70:
        continue
    print(number)

in this program out is 50,60,80,90 because we used continue funtion so when code reaches the reaches the condition it will skip thw current iterationa dn moves to next until the loop ends
