# Exception Handling


Exception handling is used to handle errors that happen while a Python program is running.Instead of allowing the program to stop suddenly, we can catch the error and display a meaningful message or take another action.

## try
try is used to test a block of code that might cause an error. If an error happens inside the try block, Python can handle it using except instead of stopping the entire program

```python
#Example:

try:
    number = int("Naveen")

except ValueError:
    print("Enter a valid number")
```

In the example python first try to execute the code inside the try,here value i have given str instead of int so its given value error 

## except
except is used to handle an error that occurs inside the try block.

```python
#Example:

try:
    number = int("Naveen")

except ValueError:
    print("Enter a valid number")
```

here int("abc") produced a ValueError.Instead of allowing the program to stop, except catches the error and displays enter a valid number

## else
else is used with try and except. The code inside else runs only when the code inside try executes successfully without any error.

```python
#Example
try:
    number = int(5)

except ValueError:
    print("Enter a valid number")
else:
    print("Number entered successfully")
```

Here what we entered in the try is the correct value to the variable so we didn't get any value error and we got else output 

## finally
finally is used with try and except. The code inside finally always runs, whether an error occurs or not.

```python
#Example:

try:
    number = int("naveen")

except ValueError:
    print("enter valid number")
finally:
    print("Programm Completed")
```

finally displays the programm is completed weather it have error or may not have error, if it have error it will eoor and display programm completed

## raise
raise is used to manually create an error in Python, Python raises an error automatically when something goes wrong. With raise, we can create an error ourselves when a specific condition is not valid.

```python
#Example:
balance = 4000
withdraw = 6000

try:
    if withdraw>balance:
        raise ValueError("Insufficent Balance")
    else:
        print("withdraw Successfull")
except ValueError as e:
    print(e)

```

In the example we maually raised the value error insufficient balance using raise keyword

## assert
assert is used to check whether a condition is true.If the condition is True → the program continues.If the condition is False → Python raises an AssertionError.

```python
#Example:

number = -5

assert number>0

print("number is positive")
```

Here the condition is not true so it given the output as assertion error

## custom exceptions
A custom exception is an error that we create ourselves for a specific situation in our program.Python already has exceptions like ValueError, TypeError, etc. But sometimes we need our own meaningful error name.

```python
#Example:
class ShortPasswordError(Exception):
    pass

password = "Naveenr"

try:
    if len(password) < 6:
        raise ShortPasswordError("Password must contain at least 6 characters")

    print("Password is valid")

except ShortPasswordError as e:
    print(e)
```

here we created our own custom exception

```python
#ValueError:

try:
    number = int("Naveen")

except ValueError:
    print("Enter a valid number")

```

- ValueError occurs when a function receives a value of the correct type, but the value itself is not valid for that operation.
- The correct way to handle a ValueError is by using a try-except block. We put the code that might produce a ValueError inside the try block. If the error occurs, Python stops executing the try block and moves to the except ValueError block. There, we can display a meaningful message or take another
    

```python
#TypeError:
try:
    name = "Naveen"
    result = name * "2"
    print(result)

except TypeError:
    print("You cannot multiply a string by another string")
```

- TypeError occurs when we use the wrong type of data for an operation or function.
- The correct way to handle this error is to use try-except and catch TypeError. When Python finds that the operation is being performed with incompatible data types, it moves to the except block, where we can display a clear message or handle the situation appropriately

```python
#IndexError:

number = [10,20,30]

try:
    print(number[3])
except IndexError:
    print("Index does not exist")
```

- IndexError occurs when we try to access an index that does not exist in a list, tuple, or other indexed collection.
    
- The correct way to handle an IndexError is to use a try-except block around the code that accesses the index. If the requested index does not exist, Python moves to the except IndexError block, where we can display a meaningful message or take another appropriate action instead of allowing the program to stop.

```python
#KeyError:

student = {"name" : "naveen",
           "age" : 24}

try:
    print(student["city"])

except KeyError:
    print("key does not exist")
    
```

- KeyError occurs when we try to access a key that does not exist in a dictionary.

- The correct way to handle a KeyError is to use a try-except block when accessing a dictionary key that might not exist. If Python cannot find the requested key, it moves to the except KeyError block.

```python
#AttributeError;

numbers = [10, 20, 30]

try:
    numbers.upper()

except AttributeError:
    print("This method is not available for a list")
```

- AttributeError occurs when we try to use a method or attribute that does not exist for a particular object.

- The correct way to handle an AttributeError is to use a try-except block around the code that might access an unavailable method or attribute. If the object does not have that method or attribute, Python moves to the except AttributeError block, where we can display a meaningful message or handle the problem appropriately.

```python
# NameError:

try:
    print(total)

except NameError:
    print("The variable 'total' is not defined")
```

- NameError occurs when we try to use a variable or function that has not been defined.

- The correct way to handle a NameError is to make sure the variable or function is defined before using it. If there is a possibility that the name may not be defined, we can use try-except to catch the error and display a clear message.

```python
# FileNotFoundError:
try:
    file = open("notes.txt", "r")
    print(file.read())

except FileNotFoundError:
    print("File not found")
```

- FileNotFoundError occurs when Python tries to open or access a file that does not exist at the given location.

- The correct way to handle FileNotFoundError is to use a try-except block when opening a file that may not exist. If Python cannot find the file, it moves to the except FileNotFoundError block, where we can display a meaningful message or take another action, such as asking the user to check the file name or location.

```python
# ZeroDivisionError:

a = 3
try:
    b = a/0
    print(a)

except ZeroDivisionError:
    print("Please enter a number other than zero")
```

- ZeroDivisionError occurs when we try to divide a number by zero. In Python, dividing by zero is not allowed, so Python raises this exception.
- The correct way to handle ZeroDivisionError is to use a try-except block around the division operation. If the denominator is zero, Python raises the exception and moves to the except ZeroDivisionError block, where we can display a meaningful message or ask the user to provide a non-zero value.    

```python
# ImportError:
try:
    from math import addition

except ImportError:
    print("The addition function is not available in the math module")
```

- ImportError occurs when Python finds a module but cannot import a specific function, class, or variable from that module.
- The correct way to handle an ImportError is to use try-except around the import statement. If the requested item cannot be imported, Python moves to the except ImportError block, where we can display a meaningful message or use an alternative.

```python
#ModuleNotFoundError:
try:
    import pandas

    print("Pandas imported successfully")

except ModuleNotFoundError:
    print("Pandas is not installed")
```

- ModuleNotFoundError occurs when Python tries to import a module that is not available in the current Python environment.
- The correct way to handle ModuleNotFoundError is to use a try-except block around the import statement. If the module is not available, the except ModuleNotFoundError block catches the error and allows us to tell the user what is missing. For an external package such as pandas, we can then install it using pip.
