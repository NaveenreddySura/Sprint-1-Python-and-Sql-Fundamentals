Arguments are values that we pass to a function when calling it. They allow the same function to work with different values.

## Default Arguments
A default argument is a value that is automatically used when we don't provide a value for that argument.

```python
#Example:

def greet(name, country="India"):
    print(name,"is from",country)

greet("Naveen")
greet("Naveen","USA")
```

In the example we can see if we don't give any value to the country it will automatically gives the default output but if we give value it prints with value which we have given

## Keyword Arguments
Keyword arguments allow us to pass values by using the parameter names. Because of this, we don't have to depend on the order of the arguments.

```python
#Example:

def student(name, age, city):
    print("Name:", name)
    print("Age:", age)
    print("City:", city)

student(city="Bengaluru",name="Naveen",age=24)
```

In the example we have give the valuus not in a order but still out the output in order because values are give to the parameters so the values will be assigned to the variables

## Variable-Length Arguments — (*args,*kwargs)

Varible Lenth Arguments allows function to allow many Arguments

*args - it allows fuction to accept any number of positional arguments
*kwargs - allows a function to accept any number of keyword arguments.

```python
#Example for *args:

def add(*numbers):
    print(numbers)

add(10,50,100,4)
```

```python
#Example for *kwargs:

def details(**information):
    print("Name:", information["name"])
    print("Age:", information["age"])
    print("City:", information["city"])

details(name="Naveen", age=24, city="Bengaluru")
```

## Recursive Functions

A recursive function keeps calling itself to solve a problem step by step until it reaches a stopping condition.

```python
#Example:

def battery_percentage(battery):
    if battery == 0:
        print("Battery empty!")
        return

    print("Battery:", battery, "%")
    battery_percentage(battery - 1)


battery_percentage(5)
```

Here the battery_percentage called itself and function stopped after reaching the stopping condition

# Iterators

An iterator is an object that allows us to access items one at a time and We can create an iterator using iter() and get the next value using next().

```python
#Example:

products = ["Laptop", "Mouse", "Keyboard"]

cart = iter(products)

print(next(cart))
print(next(cart))
```

By using Iterators we can access the values one after another by using next()

## Generator

A generator is a special type of function that produces values one at a time instead of creating all the values at once.

```python
#Example:

def count():
    for number in range(1, 6):
        yield number

for number in count():
    print(number)
```

Here the generator function given the coundown from 1 to 5 , it created one value at a time and again it create the next

## Yield
yield is used inside a generator to produce a value and pause the function.  The function resumes from where it stopped when called again.

```python
#Example:

def shopping():
    yield "Laptop"
    yield "Mouse"
    yield "Keyboard"

cart = shopping()

print(next(cart))
print(next(cart))
print(next(cart))
```

we used the yeild function inside the generator it gives the output one after another

## Decorators

A decorator is used to add extra functionality to an existing function without changing the original function's code.

```python
#Example:

def start_message(function):
    def wrapper():
        print("Starting...")
        function()
    return wrapper

@start_message
def greet():
    print("Hello Naveen")

greet()
```

## Closures
A function remembers a variable from the function where it was created, even after that outer function has finished running.

```python
#Example:

def create_greeting(name):

    def greeting():
        print("Hello", name)

    return greeting

hello = create_greeting("Naveen")

hello()
```

Here the function remembers the remebers the value from outer function

## Global and Nonlocal
global allows a function to modify a variable defined outside all functions (global scope). nonlocal allows an inner function to modify a variable in its enclosing function.




```python
#Example:

count = 0
def increase():
    global count
    count += 1

increase()
increase()

print(count)
```

Here global count allows the function to modify the global count variable and we can modify multiple times

# Built-in Functions

## map()
map() is used to apply the same function to every item in a collection.

```python
#Example:

marks = [75, 85, 95]

def add_Extramarks(final_marks):
    return final_marks + 5

new_marks = map(add_Extramarks, marks)

print(list(new_marks))
```

here it added the given condition add 5 marks to the every value of the variable

## filter()
filter() is used when we want to select only the items that satisfy a condition.

```python
#Example:

numbers = [1, 2, 3, 4, 5, 6]

even_numbers = filter(lambda number: number % 2 == 0,numbers)

print(list(even_numbers))
```

here it only given the output which only satisfied the condition

## zip()
zip() is used to combine items from two or more collections based on their position.

```python
#Example:
names = ["Naveen", "Rahul", "Priya"]
marks = [85, 90, 78]

result = zip(names, marks)

print(list(result))
```

Here we combined two different variables using zip() function

## enumerate()
enumerate() is used when we want to loop through a collection and get both the position (index) and the value.

```python
#Example:
students = ["Naveen", "Rahul", "Vimal"]

for index, student in enumerate(students):
    print(index, student)
```

Here in the output we got both index and value at a same time by using enumerate

## sorted()
sorted() is used to arrange items in a particular order.

```python
#Example:
prices = [54, 1, 78, 99, 34]
print(sorted(prices))
```

## any()
any() is used to check whether at least one item in a collection is True.

```python
#Example:
marks = [65, 90, 75, 80]

result = any(mark > 80 for mark in marks)

print(result)
```

In the example condtion checked any value >80 using any function and it has bigger value than 80 so it given true as the result

## all()
all() checks whether all items in a collection are True.

```python
#Example:

marks = [65, 90, 75, 80]

result = any(mark > 50 for mark in marks)

print(result)
```

In the example the all values satisfied the condition so output is true
